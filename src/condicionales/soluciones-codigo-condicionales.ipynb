```python
# ejercicio-final-02

c_notas = 0
notas = []

while c_notas != 3:
    c_notas += 1
    nota = float(input(f"Ingresa la nota {c_notas}: "))
    notas.append(nota)

promedio = sum(notas) / len(notas)

if promedio >= 7:
    print(f"¡Promocionado!, su promedio es de {promedio:.2f}")
else:
    print(f"No fue aceptado, su promedio es de {promedio:.2f}")

# No Promocionado:
# Ingresa la nota 1: 6.78
# Ingresa la nota 2: 7.21
# Ingresa la nota 3: 5.25
# No fue aceptado, su promedio es de 6.41

# Promocionado:
# Ingresa la nota 1: 6.77
# Ingresa la nota 2: 5.91
# Ingresa la nota 3: 9.34
# ¡Promocionado!, su promedio es de 7.34
```

```python
# ejercicio-final-06

while True:
    try:
        numero = int(input("Escriba su número: "))
        
        if numero < 0 or numero > 999:
            raise ValueError
        else:
            if numero < 10:
                print(f"El número {numero} tiene una cifra")
            elif numero >= 10 and numero <= 99:
                print(f"El número {numero} tiene dos cifra")
            else:
                print(f"El número {numero} tiene tres cifras")
    
        opcion = input("¿Desea continuar? [Sí / No]: ").strip().lower()
        
        if opcion[0] == 's':
            continue
        else:
            break

    except ValueError:
        print(f"El número {numero} no es un entero válido de tres cifras")
```

```python
# ejercicio-final-14

try:
    num1 = int(input("Ingrese el primer número: "))
    num2 = int(input("Ingrese el segundo número: "))
    num3 = int(input("Ingrese el tercer número: "))

    minimo = num1

    if num2 < minimo:
        minimo = num2

    if num3 < minimo:
        minimo = num3

    maximo = num1

    if num2 > maximo:
        maximo = num2

    if num3 > maximo:
        maximo = num3

    print(f"El número menor es: {minimo}")
    print(f"El número mayor es: {maximo}")

except ValueError:
    print("Ingrese solo números enteros válidos.")
```

```python
# ejercicio-final-15

import datetime # Necesario para obtener la hora actual

print("\n--- MENÚ PRINCIPAL ---")
print("1. Saludar")
print("2. Decir la hora")
print("3. Salir")
print("----------------------")

opcion = input("Ingrese el número de su opción (1, 2 o 3): ")

match opcion:
    case '1':
        print("¡Hola, querido usuario!")
    case '2':
        # Mostrar la hora actual
        hora_actual = datetime.datetime.now().strftime("%H:%M:%S")
        print(f"La hora actual es: {hora_actual}")
    case '3':
        print("Saliendo del programa. ¡Hasta luego!")
    case _:
        print("Opción no válida. Por favor, ingrese solo 1, 2 o 3.")
```
