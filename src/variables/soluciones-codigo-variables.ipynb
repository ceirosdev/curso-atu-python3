```python
# ejercicio60

opcion="si"
suma=0
while opcion=="si":
    valor=int(input("Ingrese un valor:"))
    suma=suma+valor
    opcion=input("Desea cargar otro numero (si/no):")
print("La suma de valores ingresados es")
print(suma)
```

```python
# ejercicio-final-01

persona1 = input('Primera persona: ').strip()
persona2 = input('Segunda persona: ').strip()

personas_ordenadas = sorted([persona1, persona2])

orden = 0

for persona in personas_ordenadas:
    orden += 1
    print(f'{orden}: {persona}')
```
