```python
# ejercicio325

lista1 = [8, 5, 4, 10, 2]
lista2 = []
for elemento in lista1:
    lista2.append(elemento * elemento)
print("Lista 1")
print(lista1)
print("Nueva lista")
print(lista2)
```

```python
# ejercicio326

lista1 = [8, 5, 4, 10, 2]
lista2 = [elemento * elemento for elemento in lista1]
print("Lista 1")
print(lista1)
print("Nueva lista")
print(lista2)
```

```python
# ejercicio327

personas = [("pedro", 33), ("ana", 3), ("juan", 13), ("carla", 45)]
personas_mayores = [per for per in personas if per[1] >= 18]
print(personas_mayores)
```

```python
# ejercicio328

multiplos8 = [valor for valor in range(1, 501) if valor % 8 == 0]
print(multiplos8)
```

```python
# ejercicio329

nombres = ["juan", "pablo", "luis", "mauro", "hector"]
nombres_compuestos = [
    [nombre1, nombre2]
    for nombre1 in nombres
    for nombre2 in nombres
    if nombre1 != nombre2
]
print(nombres_compuestos)
```
