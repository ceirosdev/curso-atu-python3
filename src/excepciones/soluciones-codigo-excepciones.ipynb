```python
# ejercicio296

while True:
    valor1=int(input("Ingrese primer valor:"))
    valor2=int(input("Ingrese segundo valor:"))
    suma=valor1+valor2
    print("La suma es", suma)
    respuesta=input("Desea ingresar otro par de valores?[s/n]")
    if respuesta=="n":
        break
```

```python
# ejercicio297

while True:
    try:
        valor1=int(input("Ingrese primer valor:"))
        valor2=int(input("Ingrese segundo valor:"))
        suma=valor1+valor2
        print("La suma es",suma)
    except ValueError:
        print("debe ingresar números.")
    respuesta=input("Desea ingresar otro par de valores?[s/n]")
    if respuesta=="n":
        break
```

```python
# ejercicio298

try:
    valor1=int(input("Ingrese dividendo:"))
    valor2=int(input("Ingrese divisor:"))
    division=valor1/valor2
    print("El resultado de la división es", division)
except ZeroDivisionError:
    print("No se puede dividir por cero.")
```

```python
# ejercicio299

meses=("enero","febrero","marzo","abril","mayo","junio",
       "julio","agosto","septiembre","octubre","noviembre","diciembre")
try:
    nromes=int(input("Ingrese un número de mes [1-12]:"))
    print(meses[nromes-1])
except IndexError:
    print("En número de mes debe ir entre 1 y 12")
```

```python
# ejercicio300

try:
    valor1=int(input("Ingrese dividendo:"))
    valor2=int(input("Ingrese divisor:"))
    division=valor1/valor2
    print("El resultado de la división es", division)
except ZeroDivisionError:
    print("No se puede dividir por cero.")
except ValueError:
    print("Los valores a ingresar deben ser enteros")
```

```python
# ejercicio301

try:
    archi1=open("datos.txt","w")
    archi1.write("Primer línea.\n")
    archi1.write("Segunda línea.\n")
    archi1.write("Tercer línea.\n")
    archi1.write(3334)
except TypeError:
    print("No se puede grabar un entero con write")
finally:
    archi1.close()
    print("Se cerró el archivo")
```
