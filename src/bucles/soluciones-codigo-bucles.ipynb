```python
# ejercicio-final-01

# Lectura de la cantidad de triángulos
try:
    n = int(input("Ingrese la cantidad de pares de datos (n): "))
except ValueError:
    print("Error: Debe ingresar un número entero.")
    exit()

# Inicialización del contador para la condición b)
contador_superficie_mayor_a_12 = 0


# Bucle for para procesar cada triángulo
for i in range(n):
    print(f"--- Procesando Triángulo N° {i + 1} ---")

    # Lectura de datos (base y altura)
    try:
        base = float(input(f"Ingrese la medida de la base para el triángulo {i + 1}: "))
        altura = float(input(f"Ingrese la medida de la altura para el triángulo {i + 1}: "))
    except ValueError:
        print("Error: Ingrese valores numéricos válidos. Saltando este triángulo.")
        continue

    # Cálculo de la superficie (Área = (base * altura) / 2)
    superficie = (base * altura) / 2

    # a) Informar los datos de cada triángulo
    print("\n--- Resultados del Triángulo ---")
    print(f"     Base: {base}")
    print(f"     Altura: {altura}")
    print(f"     Superficie (Área): {superficie:.2f}")

    # b) Verificar la condición de superficie > 12
    if superficie > 12:
        contador_superficie_mayor_a_12 += 1


# Informar el resultado final
print(f"\nCantidad de triángulos con superficie > 12: {contador_superficie_mayor_a_12}")
```

```python
# ejercicio-final-05

# Inicialización de contadores para cada tipo de triángulo
contador_equilatero = 0
contador_isosceles = 0
contador_escaleno = 0

# Solicitud de la cantidad de triángulos (n)
try:
    n = int(input("Ingrese la cantidad de triángulos a procesar (n): "))
except ValueError:
    print("Error: Debe ingresar un número entero para la cantidad.")
    exit()

# Procesar cada uno de los 'n' triángulos
for i in range(n):
    print(f"\n--- Triángulo N° {i + 1} ---")

    # Lectura de los tres lados del triángulo
    try:
        lado_a = float(input("Ingrese la longitud del lado A: "))
        lado_b = float(input("Ingrese la longitud del lado B: "))
        lado_c = float(input("Ingrese la longitud del lado C: "))

    except ValueError:
        print("Error: Ingrese valores numéricos válidos para los lados. Saltando este triángulo.")
        continue  # Saltar al siguiente ciclo si hay un error de entrada

    # 1. Clasificación del triángulo
    tipo = ""

    # Condición - Equilátero: Los tres lados son iguales
    if lado_a == lado_b and lado_b == lado_c:
        tipo = "Equilátero"
        contador_equilatero += 1

    # Condición - Isósceles: Exactamente dos lados son iguales
    elif lado_a == lado_b or lado_b == lado_c or lado_a == lado_c:
        tipo = "Isósceles"
        contador_isosceles += 1

    # Condición - Escaleno: Ningún lado es igual
    else:
        tipo = "Escaleno"
        contador_escaleno += 1

    # 2. Informar los resultados para el triángulo actual (Requisito a)
    print("\tClasificación del Triángulo:")
    print(f"   Lados ingresados: {lado_a}, {lado_b}, {lado_c}")
    print(f"   Tipo de triángulo: {tipo}")


print("\tCantidad de triángulos por tipo:")
print(f"   Equilátero: {contador_equilatero}")
print(f"   Isósceles: {contador_isosceles}")
print(f"   Escaleno: {contador_escaleno}")
```

```python
# ejercicio-final-06

# Inicialización de contadores para cada cuadrante
contador_q1 = 0  # x > 0, y > 0
contador_q2 = 0  # x < 0, y > 0
contador_q3 = 0  # x < 0, y < 0
contador_q4 = 0  # x > 0, y < 0

# --- Paso 1: Solicitar la cantidad de puntos (n) con validación continua ---
while True:
    try:
        # Pedir la cantidad de puntos
        n = int(input("Ingrese la cantidad de puntos a procesar (n): "))
        # Si la conversión es exitosa, salimos del bucle de validación
        break
    except ValueError:
        # Si hay un error (no es un número), avisamos y el bucle se repite
        print("Error: Por favor, ingrese un número entero válido.")

# --- Paso 2: Bucle for para procesar los 'n' puntos ---
for i in range(n):
    print(f"\n--- Ingresando Punto N° {i + 1} ---")

    # Lectura de las coordenadas (x, y) con validación
    try:
        x = float(input("Ingrese la coordenada X: "))
        y = float(input("Ingrese la coordenada Y: "))
    except ValueError:
        print("Error: Entrada inválida. Por favor, ingrese números.")
        continue  # Saltar al siguiente punto si la entrada es incorrecta

    # 3. Clasificación del cuadrante
    
    if x > 0 and y > 0:
        contador_q1 += 1
        print("Clasificado como Cuadrante I.")

    elif x < 0 and y > 0:
        contador_q2 += 1
        print("Clasificado como Cuadrante II.")

    elif x < 0 and y < 0:
        contador_q3 += 1
        print("Clasificado como Cuadrante III.")

    elif x > 0 and y < 0:
        contador_q4 += 1
        print("Clasificado como Cuadrante IV.")

    else:
        # Manejo de puntos sobre los ejes (x=0 o y=0)
        print("El punto se encuentra sobre un eje y no se cuenta en los cuadrantes.")


print(f"\nPuntos en el Cuadrante I (x>0, y>0): {contador_q1}")
print(f"Puntos en el Cuadrante II (x<0, y>0): {contador_q2}")
print(f"Puntos en el Cuadrante III (x<0, y<0): {contador_q3}")
print(f"Puntos en el Cuadrante IV (x>0, y<0): {contador_q4}")
```

```python
# ejercicio-final-07

# Inicialización de contadores y acumulador
negativos_count = 0
positivos_count = 0
multiplos_15_count = 0
suma_pares = 0

# Bucle for para procesar exactamente 10 valores
for i in range(10):
    try:
        # Solicitamos el valor al usuario y lo convertimos a entero
        valor = int(input(f"Ingrese el valor {i + 1} de 10: "))

        # a) Contar negativos
        if valor < 0:
            negativos_count += 1

        # b) Contar positivos
        elif valor > 0:
            positivos_count += 1

        # c) Contar múltiplos de 15
        if valor % 15 == 0:
            multiplos_15_count += 1

        # d) Acumular números pares
        if valor % 2 == 0:
            suma_pares += valor

    except ValueError:
        print("Error: Entrada inválida. Por favor, ingrese un número entero.")
        # Si hay un error de entrada, simplemente saltamos a la siguiente iteración del bucle
        continue

# a) Cantidad de negativos
print(f"a) Cantidad de valores negativos: {negativos_count}")

# b) Cantidad de positivos
print(f"b) Cantidad de valores positivos: {positivos_count}")

# c) Cantidad de múltiplos de 15
print(f"c) Cantidad de múltiplos de 15: {multiplos_15_count}")

# d) Valor acumulado de los números pares
print(f"d) Suma acumulada de los números pares: {suma_pares}")
```

```python
# ejercicio-final-08

edades_mañana = []
print("--- Ingreso de Edades Turno Mañana (5 estudiantes) ---")
for i in range(5):
    try:
        edad = int(input(f"Ingrese la edad del estudiante Mañana N° {i + 1}: "))
        edades_mañana.append(edad)
    except ValueError:
        print("Error: Ingrese un número entero válido. Se omitirá esta edad.")

# Cálculo del promedio para el Turno Mañana
suma_mañana = sum(edades_mañana)
promedio_mañana = suma_mañana / 5

edades_tarde = []
print("\n--- Ingreso de Edades Turno Tarde (6 estudiantes) ---")
for i in range(6):
    try:
        edad = int(input(f"Ingrese la edad del estudiante Tarde N° {i + 1}: "))
        edades_tarde.append(edad)
    except ValueError:
        print("Error: Ingrese un número entero válido. Se omitirá esta edad.")

# Cálculo del promedio para el Turno Tarde
suma_tarde = sum(edades_tarde)
promedio_tarde = suma_tarde / 6

edades_noche = []
print("\n--- Ingreso de Edades Turno Noche (11 estudiantes) ---")
for i in range(11):
    try:
        edad = int(input(f"Ingrese la edad del estudiante Noche N° {i + 1}: "))
        edades_noche.append(edad)
    except ValueError:
        print("Error: Ingrese un número entero válido. Se omitirá esta edad.")

# Cálculo del promedio para el Turno Noche
suma_noche = sum(edades_noche)
promedio_noche = suma_noche / 11

print(f"Promedio Turno Mañana: {promedio_mañana:.2f}")
print(f"Promedio Turno Tarde: {promedio_tarde:.2f}")
print(f"Promedio Turno Noche: {promedio_noche:.2f}")

# Determinar cuál promedio es el mayor usando sentencias condicionales
if promedio_mañana > promedio_tarde and promedio_mañana > promedio_noche:
    print("\nEl turno de Mañana tiene el promedio de edades más alto.")
elif promedio_tarde > promedio_mañana and promedio_tarde > promedio_noche:
    print("El turno de Tarde tiene el promedio de edades más alto.")
else:
    print("El turno de Noche tiene el promedio de edades más alto, o hay un empate entre Mañana y Tarde.")
```
