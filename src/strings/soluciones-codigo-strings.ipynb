```python
# ejercicio64

mail=input("Ingrese un mail:")
cantidad=0
x=0
while x<len(mail):
    if mail[x]=="@":
        cantidad=cantidad+1
    x=x+1
if cantidad==1:
    print("Contiene solo un caracter @ el mail ingresado")
else:
    print("Incorrecto")
```

```python
# ejercicio-final-01

# Escribe: "Esta oración tiene cuatro espacios"
oracion = input('Oración 1: ').strip()

cantidad_espacios = 0

for caracter in oracion:
    if caracter == " ":
        cantidad_espacios += 1

if cantidad_espacios == 0:
    print("Escribiste corrido (sin espacios)")
else: 
    print(f"La oración tiene {cantidad_espacios} espacios")
```

```python
# ejercicio-final-02

# Escribe: "Hay 4 vocales"
oracion = input("Escribe una oración: ").strip().lower()

contador = 0

for caracter in oracion:
    if caracter in ['a', 'e', 'i', 'o', 'u']: 
        contador += 1

if contador == 0:
    print("No tiene vocales tu oración")
else:
    print(f"Tu oración tiene {contador} vocales")
```
