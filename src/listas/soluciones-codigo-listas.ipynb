```python
# ejercicio-final-05

def obtener_alturas():
    alturas = []
    print("Por favor, ingrese las 5 alturas de las personas (valores flotantes):")

    for i in range(5):
        while True:
            try:
                # Solicitar la entrada del usuario
                entrada = input(f"Altura de la persona {i + 1}: ")
                # Convertir la entrada a float
                altura = float(entrada)
                alturas.append(altura)
                break  # Salir del bucle 'while' si la conversión es exitosa
            except ValueError:
                # Manejar el error si el usuario ingresa algo que no es un número
                print("Error: Por favor, ingrese un valor numérico válido.")
    
    return alturas

def analizar_alturas(alturas):
    if not alturas:
        print("No hay datos para analizar.")
        return

    # Obtener el promedio
    suma_total = sum(alturas)
    promedio = suma_total / len(alturas)

    # Contar personas más altas y más bajas
    taller_contador = 0
    bajas_contador = 0

    for altura in alturas:
        if altura > promedio:
            taller_contador += 1
        elif altura < promedio:
            bajas_contador += 1

    # Imprimir los resultados
    print("--- Resultados del Análisis ---")
    print(f"Lista de alturas ingresadas: {alturas}")
    print(f"El promedio de altura es: {promedio:.2f}")
    print(f"Personas más altas que el promedio: {taller_contador}")
    print(f"Personas más bajas que el promedio: {bajas_contador}")


if __name__ == "__main__":
    # Cargar los datos
    alturas_personales = obtener_alturas()

    # Analizar y mostrar los resultados
    analizar_alturas(alturas_personales)
```

```python
# ejercicio-final-08

def gestionar_productos():
    nombres = []
    precios = []
    NUM_PRODUCTOS = 5

    print("--- Carga de Productos ---")

    # Crear y cargar las dos listas paralelas
    for i in range(NUM_PRODUCTOS):
        # Cargar nombre
        nombre = input(f"Ingrese el nombre del producto {i + 1}: ")
        nombres.append(nombre)

        # Cargar precio (con validación de entrada numérica)
        while True:
            try:
                precio = float(input(f"Ingrese el precio del producto '{nombre}': "))
                precios.append(precio)
                break
            except ValueError:
                print("Error: Por favor, ingrese un valor numérico válido para el precio.")

    # Mostrar las listas
    print("--- LISTAS ALMACENADAS ---")
    print(f"Nombres de los productos: {nombres}")
    print(f"Precios de los productos: {precios}")

    # Definir el precio del primer producto
    if not precios:
        print("Error: La lista de precios está vacía. No se puede realizar el análisis.")
        return

    primer_precio = precios[0]
    print(f"\nEl precio del primer producto es: {primer_precio:.2f}")

    # Contar productos con precio mayor
    contador_mayores = 0
    for precio in precios:
        if precio > primer_precio:
            contador_mayores += 1

    # Mostrar el resultado final
    print("--- Resultado del Análisis ---")
    print(f"Cantidad de productos con precio mayor al primer producto: {contador_mayores}")


if __name__ == "__main__":
    gestionar_productos()
```

```python
# ejercicio-final-09

def procesar_notas():
    nombres = []
    notas = []
    NUM_ALUMNOS = 4

    print("--- INGRESO DE DATOS ---")
    # Ingresar nombre y nota de cada alumno (dos listas paralelas)
    for i in range(NUM_ALUMNOS):
        # Ingreso del nombre
        nombre = input(f"Ingrese el nombre del alumno {i + 1}: ")
        nombres.append(nombre)

        # Ingreso de la nota (con validación de entrada numérica)
        while True:
            try:
                nota = float(input(f"Ingrese la nota de {nombre}: "))
                notas.append(nota)
                break  # Salir del bucle si la entrada es válida
            except ValueError:
                print("Error: Por favor, ingrese una nota numérica válida.")

    print("--- REPORTE DE ALUMNOS ---")

    # Realizar un listado con nombres, notas y condición
    for nombre, nota in zip(nombres, notas):
        condicion = ""
        if nota >= 8:
            condicion = "Muy Bueno"
        elif 4 <= nota <= 7:
            condicion = "Bueno"
        else: # Nota < 4
            condicion = "Insuficiente"

        print(f"Alumno: {nombre:<15} | Nota: {nota:.2f} | Condición: {condicion}")

    # Imprimir cuántos alumnos tienen la leyenda “Muy Bueno”
    muy_buenos = 0
    for nota in notas:
        if nota >= 8:
            muy_buenos += 1

    print(f"\nTotal de alumnos con la leyenda 'Muy Bueno': {muy_buenos}")

if __name__ == "__main__":
    procesar_notas()
```

```python
# ejercicio-final-10

def sumar_listas():
    LISTA_SIZE = 4
    lista_a = []
    lista_b = []

    print("--- Carga de la Lista 1 (4 enteros) ---")
    for i in range(LISTA_SIZE):
        while True:
            try:
                valor = int(input(f"Ingrese el elemento {i + 1} de la Lista 1: "))
                lista_a.append(valor)
                break
            except ValueError:
                print("Error: Por favor, ingrese un número entero válido.")

    print("\n--- Carga de la Lista 2 (4 enteros) ---")
    for i in range(LISTA_SIZE):
        while True:
            try:
                valor = int(input(f"Ingrese el elemento {i + 1} de la Lista 2: "))
                lista_b.append(valor)
                break
            except ValueError:
                print("Error: Por favor, ingrese un número entero válido.")

    # Generar la tercera lista (suma de elementos en la misma posición)
    # Usamos comprensión de listas para hacer esto de forma concisa.
    lista_suma = [lista_a[i] + lista_b[i] for i in range(LISTA_SIZE)]

    # Mostrar la tercera lista
    print("--- RESULTADO ---")
    print(f"Lista A ingresada: {lista_a}")
    print(f"Lista B ingresada: {lista_b}")
    print("\nLista de sumas (Posición i de Lista A + Lista B):")
    print(lista_suma)

if __name__ == "__main__":
    sumar_listas()
```

```python
# ejercicio-final-14

def ordenar_datos_países():
    # Crear y cargar las listas paralelas
    nombres = ["España", "México", "Argentina", "Chile", "Perú"]
    # Valores de habitantes
    habitantes = [47000000, 128000000, 45000000, 19000000, 33000000]

    print("--- DATOS INICIALES ---")
    print(f"Nombres: {nombres}")
    print(f"Habitantes: {habitantes}")

    # --- ORDENAMIENTO POR NOMBRE (ALFABÉTICO) ---

    # Crear una copia de las listas para no modificar los datos originales si fuera necesario,
    # aunque aquí modificaremos la lista 'nombres' directamente.
    nombres_ordenados = sorted(nombres)
    habitantes_ordenados_por_nombre = [habitantes[i] for i in range(len(nombres_ordenados))]

    print("\n--- ORDENAMIENTO ALFABÉTICO (Por Nombre) ---")
    print(f"Nombres ordenados: {nombres_ordenados}")
    print(f"Habitantes correspondientes: {habitantes_ordenados_por_nombre}")

    # --- ORDENAMIENTO POR HABITANTES (DE MAYOR A MENOR) ---

    # Ordenar la lista de habitantes en orden descendente (mayor a menor)
    habitantes_ordenados_por_pop = sorted(habitantes, reverse=True)

    # Crear la lista de nombres correspondiente al ordenamiento por población
    nombres_ordenados_por_pop = [nombres[i] for i in range(len(habitantes_ordenados_por_pop))]

    print("\n--- ORDENAMIENTO POR HABITANTES (De Mayor a Menor) ---")
    print(f"Nombres correspondientes: {nombres_ordenados_por_pop}")
    print(f"Habitantes ordenados: {habitantes_ordenados_por_pop}")

if __name__ == "__main__":
    ordenar_datos_países()
```

```python
# ejercicio-final-18

def gestionar_empleados():
    # Solicitar la cantidad de empleados
    while True:
        try:
            cantidad_str = input("Ingrese la cantidad de empleados que tiene la empresa: ")
            cantidad = int(cantidad_str)
            if cantidad > 0:
                break
            else:
                print("Por favor, ingrese un número positivo.")
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número entero.")

    nombres = []
    sueldos = []

    print(f"\n--- Ingreso de datos para {cantidad} empleados ---")

    # Cargar los nombres y sueldos
    for i in range(cantidad):
        # Ingreso del nombre
        nombre = input(f"Ingrese el nombre del empleado {i + 1}: ")
        nombres.append(nombre)

        # Ingreso del sueldo (con validación de entrada numérica)
        while True:
            try:
                sueldo = float(input(f"Ingrese el sueldo de {nombre}: "))
                sueldos.append(sueldo)
                break
            except ValueError:
                print("Error: Por favor, ingrese un valor numérico válido para el sueldo.")

    print("--- LISTAS ORIGINALES ---")
    print(f"Nombres originales: {nombres}")
    print(f"Sueldos originales: {sueldos}")

    # Filtrar: Crear nuevas listas solo con empleados cuyo sueldo sea <= 10000
    nombres_filtrados = []
    sueldos_filtrados = []

    # Usamos zip para recorrer ambas listas al mismo tiempo
    for nombre, sueldo in zip(nombres, sueldos):
        # Si el sueldo es MENOR O IGUAL a 10000, lo incluimos en las listas filtradas
        if sueldo <= 10000:
            nombres_filtrados.append(nombre)
            sueldos_filtrados.append(sueldo)

    # Imprimir los resultados filtrados
    print("\n--- LISTAS FILTRADAS (Sueldo <= 10000) ---")
    print(f"Nombres filtrados: {nombres_filtrados}")
    print(f"Sueldos filtrados: {sueldos_filtrados}")


if __name__ == "__main__":
    gestionar_empleados()
```

```python
# ejercicio-final-19

def filtrar_lista():
    lista_original = []
    NUM_ELEMENTOS = 5

    print("--- Carga de 5 enteros ---")
    
    # Paso 1: Cargar la lista con 5 elementos enteros
    for i in range(NUM_ELEMENTOS):
        while True:
            try:
                valor = int(input(f"Ingrese el entero número {i + 1}: "))
                lista_original.append(valor)
                break
            except ValueError:
                print("Error: Por favor, ingrese un número entero válido.")

    print("\nLista Original cargada:", lista_original)

    # Paso 2: Generar una nueva lista con los elementos que NO son mayores o iguales a 10
    # Usamos una comprensión de listas para filtrar: solo incluimos el 'elemento' 
    # si su valor es estrictamente MENOR que 10 (< 10).
    lista_filtrada = [elemento for elemento in lista_original if elemento < 10]

    # Paso 3: Imprimir la nueva lista
    print("\n--- Lista Filtrada (Elementos < 10) ---")
    print(lista_filtrada)

if __name__ == "__main__":
    filtrar_lista()
```
