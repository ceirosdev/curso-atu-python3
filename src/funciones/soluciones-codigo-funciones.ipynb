```python
# ejercicio-final-07

"""
Confeccionar una función que calcule la superficie de un rectángulo y la 
retorne, la función recibe como parámetros los valores de dos de sus lados: 
En el bloque principal del programa cargar los lados de dos rectángulos y luego 
mostrar cual de los dos tiene una superficie mayor.
"""

# Calcula la superficie de un rectámgulo
def superficie_rectangulo(
    l_ancho: float,
    l_largo: float
) -> float:
    return l_ancho * l_largo

# Muestra el rectángulo que tiene mayor suerficie
def comparar_superficie_rectangulos() -> None:
    lados1, lados2 = [], []

    print("\tCargar lados del primer rectángulo")

    for i in range(1, 3):
        lados1.append(float(input(f"Cargar el lado {i} "
                                  + "del primer rectángulo: ")))

    print("\n\tCargar lados del segundo rectángulo")

    for i in range(1, 3):
        lados2.append(float(input("Cargar el lado "
                                  + f"{i} del segundo rectángulo: ")))

    # Calcula las superficies de los dos rectángulos
    superf_rectangulo1 = superficie_rectangulo(*lados1)
    superf_rectangulo2 = superficie_rectangulo(*lados2)

    # Muestra cuál de los dos rectángulos tienen la mayor superficie
    if superf_rectangulo1 == superf_rectangulo2:

        print("\nLos rectángulos tienen la misma superficie de "
              + f"{superf_rectangulo1:.2f} u^2.")
        
    elif superf_rectangulo1 > superf_rectangulo2:
        
        print("\nEl primer rectángulo tiene la mayor superficie de " 
              + f"{superf_rectangulo1:.2f} u^2 frente al segundo con " 
              + f"{superf_rectangulo2:.2f} u^2.")

    else:
        print("\nEl segundo rectángulo tiene la mayor superficie de " 
                      + f"{superf_rectangulo2:.2f} u^2 frente al primero con " 
                      + f"{superf_rectangulo1:.2f} u^2.")

if __name__ == "__main__":

    comparar_superficie_rectangulos()
```

```python
# ejercicio-final-10

"""
Desarrollar una función que reciba una lista de string y nos retorne el que 
tiene más caracteres. Si hay más de uno con dicha cantidad de caracteres debe 
retornar el que tiene un valor de componente más baja. En el bloque principal 
iniciamos por asignación la lista de string:
"""

def mas_caracteres(palabras: list[str]) -> str:
    palabra = ""
    cant_caracteres = 0

    for caracter in palabras:
        # Obtener la cantidad de caracteres de la palabra actual
        tam_caracter = len(caracter)

        # Guarda el caracter con mayor tamaño y su nombre
        if cant_caracteres < tam_caracter:
            cant_caracteres = tam_caracter
            palabra = caracter

    return palabra

if __name__ == "__main__":

    capitales = ["Londres", "Moscú", "La Habana", "Santo Domingo", "Sofía"]

    # Imprimirá 'Santo Domingo'
    print(mas_caracteres(capitales))
```

```python
# ejercicio-final-12

"""
En una empresa se almacenaron los sueldos de 10 personas. Desarrollar las 
siguientes funciones y llamarlas desde el bloque principal:
1) Carga de los sueldos en una lista.
2) Impresión de todos los sueldos.
3) Cuántos tienen un sueldo superior a $4000.
4) Retornar el promedio de los sueldos
5) Mostrar todos los sueldos que están por debajo del promedio.
"""

SUELDOS: list[float] = []


def cargar_sueldos():

    print("\n\tCargando los sueldos de la empresa")

    indice = 1

    while True:

        SUELDOS.append(float(input(f"Sueldo {indice}: ")))

        indice += 1

        continuar = input("¿Desea continuar? [Sí / No]: ").lower().strip()

        if continuar == "n":
            break


def mostrar_sueldos():

    print("\n\tMostrando los sueldos de la empresa")

    indice = 1

    for sueldo in SUELDOS:
        print(f"{indice}: {sueldo:.2f}")
        indice += 1


def promedio_sueldos():

    # print("\n\tMostrando el promedio de los sueldos de la empresa")

    promedio = sum(SUELDOS) / len(SUELDOS)

    # print(f"El promedio de los sueldos es de {promedio:.2f}")

    return promedio


def sueldos_debajo_del_promedio():

    print("\n\tMostrando los sueldos de la empresa por debajo del promedio")

    promedio = promedio_sueldos()
    indice = 1

    for sueldo in SUELDOS:

        if sueldo < promedio:
            print(f"{indice}: {sueldo:.2f}")
            indice += 1


if __name__ == "__main__":
    
    cargar_sueldos()
    mostrar_sueldos()

    print("\n\tMostrando el promedio de los sueldos de la empresa")

    promedio = promedio_sueldos()

    print(f"El promedio de los sueldos es de {promedio:.2f}")

    sueldos_debajo_del_promedio()
```

```python
"""
Desarrollar una aplicación que permita ingresar por teclado los nombres de 
5 artículos y sus precios. Definir las siguientes funciones:
1) Cargar los nombres de artículos y sus precios.
2) Imprimir los nombres y precios.
3) Imprimir el nombre de artículo con un precio mayor
4) Ingresar por teclado un importe y luego mostrar todos los artículos con 
un precio menor igual al valor ingresado.
"""

ARTICULOS = dict()

# 1. Cargar los nombres de artículos y precios
def nombre_precio() -> None:

    print("\n\tCargando los artículos y precios")

    indice = 1

    while True:

        nombre = input(f"Ingresa el nombre del artículo "
                       + f"{indice}: ").lower().strip()

        ARTICULOS[nombre] = float(input(f"Precio del artículo {nombre}: "))

        indice += 1

        continuar = input("\t¿Desea continuar? [Sí / No]: ").lower().strip()

        if continuar == "n":
            break

# 2. Imprimir los nombres y precios
def mostrar_nombres_y_precios():

    print("\n\tMostrando los artículos y precios")

    for n_articulo, p_articulo in ARTICULOS.items():
        print(f"Artículo: {n_articulo}; Precio {p_articulo:.2f} €")

# 3. Imprimir el nombre del artículo con un precio mayor
def articulo_mayor_precio():

    print("\n\tMostrando el artículo con mayor precio")

    nombre = ""
    precio = 0.0

    for n_articulo, p_articulo in ARTICULOS.items():
        if precio < p_articulo:
            nombre = n_articulo
            precio = p_articulo

    print(f"El artículo con mayor precio es: {nombre} con {precio:.2f} €")

# 4. Muestra todos los artículos con un precio menor al monto pasado
def informe_precios_menores(monto: float):

    print(f"\n\tMostrando los artículos con un precio menor que {monto:.2f} €")
    
    art_filtrados = dict()

    for n_articulo, p_articulo in ARTICULOS.items():

        if p_articulo < monto:
            art_filtrados[n_articulo] = p_articulo

    for n_articulo, p_articulo in art_filtrados.items():
            print(f"Artículo: {n_articulo}; Precio {p_articulo:.2f} €")
    

if __name__ == "__main__":
    nombre_precio()
    mostrar_nombres_y_precios()
    articulo_mayor_precio()
    informe_precios_menores(20)
```
